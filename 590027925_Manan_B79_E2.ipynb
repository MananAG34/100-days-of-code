1. Divisibility by 3 and 5

Check if a number is divisible by both 3 and 5 using the and operator.

```python
num = int(input("Enter a number: "))

# A number is divisible by x if the remainder (modulus) is 0
if num % 3 == 0 and num % 5 == 0:
    print(f"{num} is divisible by both 3 and 5")
else:
    print(f"{num} is not divisible by both 3 and 5")
```

5. Quadratic Equation Roots

```python
import math

a = float(input("Enter a: "))
b = float(input("Enter b: "))
c = float(input("Enter c: "))

d = (b**2) - (4*a*c)

if d >= 0:
    root1 = (-b + math.sqrt(d)) / (2*a)
    root2 = (-b - math.sqrt(d)) / (2*a)
    print(f"Real Roots: {root1} and {root2}")
else:
    print("Roots are imaginary")
```

6. Leap Year Check

```python
year = int(input("Enter year: "))

# Leap year logic: divisible by 4 but not 100, unless divisible by 400
if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} is a Leap Year")
else:
    print(f"{year} is not a Leap Year")
```

7. Next Date Calculator

```python
day = int(input("day="))
month = int(input("month="))
year = int(input("year="))

# Simplifying month lengths (ignoring leap year for basics, or add logic if needed)
if month in [1, 3, 5, 7, 8, 10, 12]:
    max_days = 31
elif month in [4, 6, 9, 11]:
    max_days = 30
else:
    # Basic Feb check
    max_days = 29 if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0) else 28

if day < max_days:
    day += 1
else:
    day = 1
    if month == 12:
        month = 1
        year += 1
    else:
        month += 1

print(f"Next Date: day={day} month={month} year={year}")
```
